#### Crear External Locations requeridas para el Proyecto
1. bronze
2. silver
3. gold

```sql
DESC EXTERNAL LOCATION sadatabricksdata01_bronze;
```

```sql
%fs
ls "abfss://bronze@sadatabricksdata01.dfs.core.windows.net/"
```

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS sadatabricksdata01_silver
    URL 'abfss://silver@sadatabricksdata01.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL `databricks-ext-storage-credential`);
```

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS sadatabricksdata01_gold
    URL 'abfss://gold@sadatabricksdata01.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL `databricks-ext-storage-credential`);
```
